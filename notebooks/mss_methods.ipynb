# Configuration Region for MSS
---

```python
n = 3
p = 0.25
k = 0.99
alpha = 1 / (1 + abs(k))

T = np.hstack([(1 - p) * np.ones((n, 1)), np.vstack([p * np.eye(n - 1), np.zeros((1, n - 1))])])
T = np.hstack([
    np.vstack([(1 - p) * np.ones((n, 1)), [0]]), 
    np.vstack([p * np.eye(n), np.append(np.zeros(n - 1), 1)])
])
C = np.append(np.ones(n), 0)


# piece-wise linear
_V = np.zeros(n + 1)
V = np.zeros(n + 1)

for _ in range(int(1e7)):
    delta = np.tile(-V + C, (n + 1, 1)).T + np.tile(V, (n + 1, 1))
    pos = delta > 0
    neg = delta < 0
    _V = V.copy()
    V = V + alpha * np.sum((1 + k) * pos * T * delta + (1 - k) * neg * T * delta, axis=1)

    error = max(abs(_V - V))
    if error < 1e-2:
        print(f'Convergiu com {_} passos')
        break
    
    print(f'Iteration: {_} | Error: {error}', end='\r')

print(_V)
print(V)

n = 5
pk = p * (1 - k)
one_pk = (1 - p) * (1 + k)
print((1 - pk**n) / (one_pk * pk**n))
```
